```python
import pandas as pd
from pathlib import Path

folder = Path("C:/Users/AW'z/Downloads/ballet_Analysis_Results/260201__analysis_metrics/26020113_analysis_metrics/")

files = {
    "energy": folder / "energy.csv",
    "geometry": folder / "geometry.csv",
    "stability": folder / "stability.csv",
    "sync": folder / "synchronization.csv",
    "trans": folder / "transition.csv"
}

dfs = {name: pd.read_csv(file) for name, file in files.items() if file.exists()}

# Merge logic from notebook
merged = dfs["energy"][["energy"]].copy()
merged["frame"] = range(len(merged))

for name in ["geometry", "stability", "sync", "trans"]:
    if name in dfs:
        df = dfs[name]
        if "frame" not in df.columns:
            df["frame"] = range(len(df))
        merged = merged.merge(df, on="frame", how="outer")

merged = merged.fillna(method="ffill").fillna(0)

metrics = ["energy", "volume", "curvature", "height", "sway", "left", "right", "correlation", "jerk"]

print("--- Numerical Ranges ---")
for m in metrics:
    if m in merged.columns:
        print(f"{m}: {merged[m].min()} ~ {merged[m].max()}")
    else:
        print(f"{m}: NOT FOUND")
```
